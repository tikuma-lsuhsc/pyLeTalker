```python
from sympy import *

rho, c, l, A = symbols("rho,c,l,A")
```

```python
Ay = 0
by = 1
L = rho / A * l
cy = 1 / L
dy = 0
Y = A / (rho * c)
```

```python
Q = Matrix([[1, -1, 1], [Y, Y, 0], [Y, 0, -dy]])
P = Matrix([[0], [0], [cy]])
R = Matrix([[1, -1], [Y, Y], [0, Y]])
```

```python
Cex = Q**-1 * P
Dex = Q**-1 * P
A = Ay + by * Cex[-1]
B = by * Dex[-1]

C = Cex[:2]
D = Dex[:2]


display(Cex[:2])
display(Dex[:2])
display(A)
display(B)
```

```python
s, tau = symbols("s,tau")

together((C[0] * ((s - A) ** -1) * B + D[0]).subs(c / l, 1 / tau), s)
```
