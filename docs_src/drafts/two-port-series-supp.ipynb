```python
from sympy import *

rho, c, l, A = symbols("rho,c,l,A")
```

```python
L = rho / A * l
Ay = 0
by = 1/L
cy = 1
dy = 0
Y = A / (rho * c)
```

```python
Q = Matrix([[1, -1, 1], [Y, Y, 0], [Y, 0, -dy]])
P = Matrix([[0], [0], [cy]])
R = Matrix([[1, -1], [Y, Y], [0, Y]])
```

```python
Cex = Q**-1 * P
Dex = Q**-1 * R
A = Matrix([[Ay]]) + by * Cex[-1:, :]
B = by * Dex[-1:, :]

display(Cex.shape)
display(Dex.shape)

C = Cex[:2, :]
D = Dex[:2, :]
 
s, tau = symbols("s,tau")

display(A.subs(c / l, 1 / tau))
display(B.subs(c / l, 1 / tau))
display(C.subs(c / l, 1 / tau))
display(D.subs(c / l, 1 / tau))
```

```python
# Series R

Zr,Y = symbols("R,Y")

Q = Matrix([[1, -1, 1], [Y, Y, 0], [Y, 0, -1/Zr]])
R = Matrix([[1, -1], [Y, Y], [0, Y]])
Dex = Q**-1 * R

display(simplify(Dex))
```
