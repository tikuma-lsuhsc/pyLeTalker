```python
from sympy import *

n1, n2 = symbols("n_1,n_2", integer=True)
A1 = MatrixSymbol(r"\mathbf{A}_1", n1, n1)
b11 = MatrixSymbol(r"\mathbf{b}_{11}", n1, 1)
b12 = MatrixSymbol(r"\mathbf{b}_{12}", n1, 1)
c11 = MatrixSymbol(r"\mathbf{c}_{11}", 1, n1)
c12 = MatrixSymbol(r"\mathbf{c}_{12}", 1, n1)
d111, d112, d121, d122 = symbols(r"d_{111},d_{112},d_{121},d_{122}")
A2 = MatrixSymbol(r"\mathbf{A}_2", n2, n2)
b21 = MatrixSymbol(r"\mathbf{b}_{21}", n2, 1)
b22 = MatrixSymbol(r"\mathbf{b}_{22}", n2, 1)
c21 = MatrixSymbol(r"\mathbf{c}_{21}", 1, n2)
c22 = MatrixSymbol(r"\mathbf{c}_{22}", 1, n2)
d211, d212, d221, d222 = symbols(r"d_{211},d_{212},d_{221},d_{222}")

B1 = BlockMatrix([[b11, b12]])
B2 = BlockMatrix([[b21, b22]])
C1 = BlockMatrix([[c11], [c12]])
C2 = BlockMatrix([[c21], [c22]])
D1 = Matrix([[d111, d112], [d121, d122]])
D2 = Matrix([[d211, d212], [d221, d222]])

U = Matrix([[1, 0], [0, 0]])
L = Matrix([[0, 0], [0, 1]])
I = Matrix([[1, 0], [0, 1]])
```

```python
display(U)
display(L)
display(I)
```

```python
Q1inv = (I - D1 * L * D2 * U) ** -1
Q2inv = (I - D2 * U * D1 * L) ** -1
P1 = D1 * (L * D2 * L + U)
P2 = D2 * (U * D1 * U + L)
Qinvden = d112 * d221 - 1
display(Q1inv * Qinvden, Q2inv * Qinvden, P1, P2)
```

```python
display(L * Q2inv)
display(U * Q1inv)
```

```python
# D matrix
Dden = Qinvden
display(simplify(simplify(U * Q2inv * P2 + L * Q1inv * P1) * Dden), Dden)
```

```python
# C matrix
display(simplify(expand((L * Q1inv + U * Q2inv * D2 * U) * C1 * Qinvden)), Qinvden)
display(simplify(expand((U * Q2inv + L * Q1inv * D1 * L) * C2 * Qinvden)), Qinvden)
```

```python
# B matrix
display(simplify(B1 * simplify((L * Q2inv * P2 + U) * Qinvden)), Qinvden)
display(simplify(B2 * simplify((U * Q1inv * P1 + L) * Qinvden)), Qinvden)
```

```python
# A matrix
display(simplify(A1 + B1 * simplify((L * Q2inv * D2 * U) * Qinvden) * C1))
display(simplify(B1 * L * simplify(Q2inv * Qinvden) * C2))
display(simplify(B2 * U * simplify(Q1inv * Qinvden) * C1))
display(simplify(A2 + B2 * simplify((U * Q1inv * D1 * L) * Qinvden) * C2), Qinvden)
```

```python
# lossless junction
Y1, Y2, A1, A2, rhoc = symbols(r"Y_1,Y_2,A_1,A_2,c\rho")
Y1 = A1 / rhoc
Y2 = A2 / rhoc
simplify(
    Matrix([[1, -1], [Y2, Y1]]) ** -1
    # * (Y1 + Y2)
    * (A1 + A2)
    * Matrix([[1, -1, -1, 0], [Y1, Y2, 0, 1]])
)
```

```python
# yielding wall
Y, dw, A, rhoc = symbols(r"Y,d_w,A,c\rho")
Y = A / rhoc
simplify(
    Matrix([[1, -1], [Y, Y + dw]]) ** -1
    # * (2 * Y + dw)
    * (2 * A + rhoc * dw)
    # * Matrix([[1, -1], [Y - dw, Y]])
)
# simplify(
#     Matrix([[1, -1], [Y, Y + dw]]) ** -1
#     * (2 * A + rhoc * dw)
#     * Matrix([[1, -1], [Y - dw, Y]])
# )
```

```python
# viscous loss
Y, dv, A, rhoc = symbols(r"Y,d_v,A,c\rho")
Y = A / rhoc
simplify(
    Matrix([[dv * Y + 1, -1], [Y, Y]]) ** -1
    # * (Y * dv + 2)
    * (A * dv + 2 * rhoc)
    * Matrix([[1, dv * Y - 1], [Y, Y]])
)
# simplify(
#     Matrix([[1, -1], [Y, Y + dw]]) ** -1
#     * (2 * A + rhoc * dw)
#     * Matrix([[1, -1], [Y - dw, Y]])
# )
```

```python
# lossless junction + yielding wall
D1 = (
    Matrix([[1, -1], [Y2, Y1]]) ** -1
    * (A1 + A2)
    * Matrix([[1, -1], [Y1, Y2]])
    / (A1 + A2)
)
D2 = (
    Matrix([[1, -1], [Y2, Y2 + dw]]) ** -1
    * (2 * A2 + rhoc * dw)
    * Matrix([[1, -1], [Y2 - dw, Y2]])
    / (2 * A2 + rhoc * dw)
)
Q1inv = (I - D1 * L * D2 * U) ** -1
Q2inv = (I - D2 * U * D1 * L) ** -1
P1 = D1 * (L * D2 * L + U)
P2 = D2 * (U * D1 * U + L)
dexpr = A1 + A2 + rhoc * dw
simplify(simplify(U * Q2inv * P2 + L * Q1inv * P1) * dexpr)
```

```python
# first D term
d2expr = 2 * A2 + rhoc * dw
d1expr = A1 + A2
den = symbols("D")
D11 = (
    Matrix([[2 * A2 / d2expr, 0], [0, 2 * A2 / d1expr]])
    * Matrix([[1, (A2 - A1) / d1expr], [-rhoc * dw / d2expr, 1]])
    * Matrix([[2 * A1 / d1expr, 0], [0, 2 * A2 / d2expr]])
    * d1expr
    * d2expr
    / (2 * A2 * den)
    * den
)
display(D11)

# add the 2nd D term
D22 = Matrix([[0, -rhoc * dw / d2expr], [(A1 - A2) / d1expr, 0]])
display(D22)

# finally
simplify(D11 + D22 * dexpr)
```

```python
cw = symbols(r"\mathbf{c}_w")
simplify((U * Q2inv + L * Q1inv * D1 * L) * (-Matrix([[1], [1]])) * dexpr)
```

```python
bw = MatrixSymbol(r"\mathbf{b}_w", 2, 1)
ones = Matrix([[1], [1]])
B2 = 2 * A2 / (2 * A2 + rhoc * dw) * bw * ones.T
simplify(B2 * (U * Q2inv * P2 + L * Q1inv * P1))
```
